12.find common values between two arrays

```python
import numpy as np

arr1 = np.array([0, 10, 20, 40, 60])
arr2 = np.array([10, 30, 40])

common_values = []
for value in arr1:
    if value in arr2:
        common_values.append(value)
common_values = np.array(common_values)

print("common values: ",common_values )
```

13.values in array1 that are not in array2...return it sorted

```python
import numpy as np

arr1 = np.array([0, 10, 20, 40, 60, 80])
arr2 = np.array([10, 30, 40, 50, 70, 90])

diff_array = []
for value in arr1:
    if value not in arr2:
        diff_array.append(value)

diff_array = np.array(sorted(diff_array))
print(diff_array)
```

14.unique values that are only in one array(not both)..return it in sorted

```python
import numpy as np

arr1 = np.array([0, 10, 20, 40, 60, 80])
arr2 = np.array([10, 30, 40, 50, 70])

res_array = []
for value in arr1:
    if value not in arr2:
        res_array.append(value)

for value in arr2:
    if value not in arr1:
        res_array.append(value)

res_array = np.array(sorted(res_array))
print(res_array)
```
